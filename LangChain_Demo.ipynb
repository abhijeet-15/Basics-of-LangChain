```python
from langchain_huggingface import HuggingFacePipeline
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
```

```python
llm = HuggingFacePipeline.from_model_id(
    model_id="Qwen/Qwen2.5-1.5B-Instruct",
    task="text-generation",
    pipeline_kwargs= {
        "max_new_tokens": 100
    }
)
```

```python
prompt = "What is the time complexity of binary search in an unsorted array?"
```

```python
response = llm.invoke(prompt)
print(response)
```

```python
algo_tutor_template = """
You are an expert Computer Science engineer. Explain the following algorithm: {algorithm_name}

Follow the EXACT structure:

### 1. The concept(what is it?)
Explain the problem that this algorithm solves in max 2 sentences.

### 2. Complexity analysis
Define the time and space complexity of the {algorithm_name}

### 3. Sample trace run
Show a sample dry-run on the provided input: {sample_input}

Answer::
"""
```

```python
prompt = PromptTemplate(
    template=algo_tutor_template,
    input_variables=["algorithm_name", "sample_input"]
)
```

```python
parsed_prompt = prompt.format(
    algorithm_name = "Bubble sort",
    sample_input = [1, 99, 1 , 12, 23, 4]
)

print(parsed_prompt)
```

```python
chain = prompt | llm
```

```python
result = chain.invoke({
    "algorithm_name": "Binary search",
    "sample_input" : "Target 8 in the list [1,7,10,34]"
})
```

```python
template = ChatPromptTemplate.from_messages([
    ("system", "You are a calculator that responds with reasoning about the calculation"),
    ("human", "What is 10 multiplied by 2"),
    ("ai", "10X2=20"),
    ("human", "{question}")
])
```

```python
chain = template | llm
response = chain.invoke({
    "question": "What is square root of 10, upto 3 decimal places"
})
```

```python
print(response)
```
